```python
import os

# Path to dataset relative to notebook folder
dataset_path = "../brain_tumor_dataset/"

# List files/folders inside dataset
print(os.listdir(dataset_path))
```

```python
import tensorflow as tf

dataset_path = "../brain_tumor_dataset/"

# Load dataset directly from folders
train_ds = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    labels='inferred',            # Automatically infer labels from folder names
    label_mode='categorical',     # Use 'categorical' for multi-class classification
    image_size=(224, 224),        # Resize all images to 224x224
    batch_size=32,                # Number of images per batch
    shuffle=True                  # Shuffle data
)

# Check a batch
for images, labels in train_ds.take(1):
    print("Batch image shape:", images.shape)
    print("Batch label shape:", labels.shape)

```

```python
from tensorflow.keras import layers, models

# Define CNN model
model = models.Sequential([
    layers.Rescaling(1./255, input_shape=(224,224,3)),  # Normalize images
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.MaxPooling2D((2,2)),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D((2,2)),
    layers.Conv2D(128, (3,3), activation='relu'),
    layers.MaxPooling2D((2,2)),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(2, activation='softmax')  # 2 classes: no tumor, yes tumor
])

# Compile model
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Train model
history = model.fit(
    train_ds,
    epochs=10  # You can increase later
)

# Save trained model
model.save('brain_tumor_model.h5')
```
